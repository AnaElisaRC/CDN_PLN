## Questão 01 (1.0)
a) Explique o que é uma representação esparsa em PLN e por que a codificação one-hot é considerada esparsa.

b) Na célula abaixo, implemente em Python a codificação one-hot para a lista ["casa", "carro", "filme", "bom", "ruim"] utilizando OneHotEncoder do sklearn e exiba a matriz resultante.

### RESPOSTA

Item (1A)

R. São vetores de alta dimensionalidade. One-hot é considerada esparsa porque quase todas as posições são 0, contendo apena um 1 por token

```
!pip install scikit-learn
```

```
# RESPOSTA Item (1B)
# Seu código abaixo:

vocab = ["casa", "carro", "filme", "bom", "ruim"]
index = {w: i for i, w in enumerate(vocab)}

def one_hot(word, vocab_index):
    vec = [0] * len(vocab_index)
    vec[vocab_index[word]] = 1
    return vec

print("Vocabulário:", index)
for w in vocab:
    print(w, "→", one_hot(w, index))

```

a) Explique o que é o modelo Bag of Words (BoW) e qual sua principal limitação em tarefas de PLN.

b) Utilizando o CountVectorizer do sklearn, aplique BoW ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"

### RESPOSTA

Item (5A)

R. O Bag of Words ignora a ordem e a gramática, focando apenas na frequência (contagem) das palavras. Sua principal limitação é a perda do contexto e o significado das frases.

```
# RESPOSTA Item (5B)
# Seu código abaixo:

from nltk.tokenize import sent_tokenize
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
import spacy
import nltk


# Exemplo de parágrafo
textos = """
O sistema falhou.
O sistema foi atualizado.
Falha no sistema.
"""

# Tokenizar em sentenças
sentencas = sent_tokenize(textos, language='portuguese')

# Remover stopwords e lematizar
corpus_sem_stop = []

for texto in sentencas:
    doc = nlp(texto)
    tokens_limpos = [
        token.lemma_
        for token in doc
        if token.is_alpha and not token.is_stop
    ]
    corpus_sem_stop.append(" ".join(tokens_limpos))

print("Corpus processado:", corpus_sem_stop)

# Criar Bag of Words
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus_sem_stop)
feature_names = vectorizer.get_feature_names_out()

# Visualizar a matriz BoW
bow_df = pd.DataFrame(X.toarray(), columns=feature_names)
print("\nVocabulário:", feature_names)
print("\nMatriz BoW:\n", bow_df)
```

### RESPOSTA

Item (7A)

R. Representa quantas vezes palavras aparecem próximas umas das outras no corpus, utilizando uma janela de contexto. Seu objetivo é identificar palavras que coocorrem significativamente, ajudando a capturar relações semânticas entre termos.

### RESPOSTA

Item (7B)

R. Gato e leite são independentes. A probabilidade de ocorrer “gato” e “leite” juntos é exatamente igual ao produto das probabilidades individuais.

```
# RESPOSTA  Item (7C)
# Seu código abaixo:

# Corpus
corpus = [
    "gato bebe agua",
    "cachorro bebe agua",
    "gato bebe leite",
    "cachorro bebe leite",
    "gato come peixe",
    "cachorro come peixe",
    "gato come carne",
    "cachorro come carne",
    "homem bebe agua",
    "homem come carne",
]

# Visualizar o corpus.
for i, frase in enumerate(corpus, 1):
    print(f"D{i}: {frase}")

import numpy as np
import pandas as pd

# Gerar bigramas sequenciais (janela de contexto = "1 à direita").
from nltk.util import bigrams

bigrams_list = []
for sent in corpus:
    tokens = sent.split()
    bigrams_list.extend(list(bigrams(tokens)))

bigrams_list

# Contar frequência dos bigramas.
from collections import Counter

bigram_freq = Counter(bigrams_list)
bigram_freq

# Criar matriz de coocorrência a partir dos bigramas.
words = sorted(set([w for bg in bigram_freq for w in bg]))
matrix = pd.DataFrame(0, index=words, columns=words)

for (w,c), freq in bigram_freq.items():
    matrix.loc[w,c] = freq

matrix

# Gerar matriz de probabilidade conjunta.
N = matrix.values.sum()
p_wc = matrix / N
p_wc

# Calcular frequência marginal das palavras.
p_w = matrix.sum(axis=1) / N
p_w

# Calcular frequência marginal dos contextos.
p_c = matrix.sum(axis=0) / N
p_c

# Calcular PMI (Pointwise Mutual Information).
pmi = np.log(p_wc / (p_w.values[:, None] * p_c.values[None, :]))
pmi = pmi.replace([np.inf, -np.inf], 0)
pmi
```
