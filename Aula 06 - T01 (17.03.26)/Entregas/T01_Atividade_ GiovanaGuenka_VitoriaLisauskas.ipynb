## Questão 01 (1.0)
a) Explique o que é uma representação esparsa em PLN e por que a codificação one-hot é considerada esparsa.

b) Na célula abaixo, implemente em Python a codificação one-hot para a lista ["casa", "carro", "filme", "bom", "ruim"] utilizando OneHotEncoder do sklearn e exiba a matriz resultante.

### RESPOSTA

Uma representação esparsa é um vetor de alta dimensionalidade no qual a maioria das posições contém zero. Em One-Hot, cada token recebe um vetor de tamanho igual ao vocabulário (V), com apenas 1 na posição correspondente ao token e zeros em todas as demais. Nesse sentido, para vocabulários grandes quase toda a informação é zero, tornando a representação esparsa e ineficiente em memória

```
# Resposta 1B
corpus = ["casa", "carro", "filme", "bom", "ruim"]
index = {w: i for i, w in enumerate(corpus)}

def one_hot(word, corpus_index):
    vec = [0] * len(corpus_index)
    vec[corpus_index[word]] = 1
    return vec

print("Corpus:", index)
for w in corpus:
    print(w, "→", one_hot(w, index))
```

## Questão 02 (1.0)

a) Explique o que é stemming e qual é seu objetivo no processamento de linguagem natural.

b) Utilizando o NLTK e o RSLPStemmer, aplique stemming ao texto abaixo:
"Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

### RESPOSTA

Stemming é a redução de palavras ao seu radical que removem sufixos e terminações. Seu objetivo é agrupar variações morfológicas de um mesmo termo sob uma base comum, reduzindo a dimensionalidade do vocabulário. No entanto, não necessariamente o radical terá valor semântico (a palavra pode não existir)

```
# RESPOSTA Item (2B)
import nltk
nltk.download('rslp', quiet=True)
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
texto = "Os pesquisadores analisavam relatórios e comparavam resultados financeiros."
tokens = texto.lower().replace(".", "").split()
stems = [(t, stemmer.stem(t)) for t in tokens]
for token, stem in stems:
    print(f"{token:<20} {stem}")
```

## Questão 03 (1.0)

a) Explique o que são stopwords em PLN e qual é o objetivo de removê-las durante o processamento de texto.

b) Utilizando o NLTK, realize a remoção de stopwords do texto abaixo e exiba os tokens restantes:

### RESPOSTA

Stopwords são palavras muito frequentes (artigos, preposições, conjunções) que contribuem pouco para diferenciar documentos. Remove-las reduz ruído estatístico e dimensionalidade, mantendo apenas termos com maior significância

```
# RESPOSTA Item (3B)

import nltk
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_pt = set(stopwords.words('portuguese'))
texto = "Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."
tokens = word_tokenize(texto.lower(), language='portuguese')
filtrados = [t for t in tokens if t.isalpha() and t not in stop_pt]
print("Tokens restantes:", filtrados)
```

## Questão 05 (1.0)

a) Explique o que é o modelo Bag of Words (BoW) e qual sua principal limitação em tarefas de PLN.

b) Utilizando o CountVectorizer do sklearn, aplique BoW ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

### RESPOSTA

BoW representa um documento como vetor de frequências de tokens no vocabulário, ignorando a ordem das palavras. Sua principal limitação é não capturar semântica: sinônimos ("carro", "automóvel") têm vetores diferentes, e não detecta que textos com palavras distintas podem tratar do mesmo tema.

```
# RESPOSTA Item (5B)

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

textos = ["o sistema falhou", "o sistema foi atualizado", "falha no sistema"]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(textos)
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out(),
                  index=[f"Doc{i+1}" for i in range(len(textos))])
print("Vocabulário:", vectorizer.get_feature_names_out().tolist())
print(df.to_string())
```

a) Explique o que é TF-IDF e qual é sua principal ideia em relação à importância das palavras em um conjunto de documentos.

b) Compare TF-IDF com Bag of Words (BoW), destacando a principal diferença entre as duas abordagens.

c) Utilizando o TfidfVectorizer do sklearn, aplique TF-IDF ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:

textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

### RESPOSTA

6a) TF-IDF pondera a frequência de um termo no documento (TF) pela raridade no corpus (IDF = log(N/DF)). A ideia central é aumentar o peso de termos informativos (raros no corpus) e reduzir o peso de termos comuns a todos os documentos. Em outras palavras, TF-IDF destaca (discrimina) quais termos são diferentes raros

### RESPOSTA

6b)
| Aspecto | BoW | TF-IDF |
| :-- | :-- | :-- |
| Peso | Contagem bruta | Frequência × raridade |
| Termos comuns | Peso alto | Peso próximo de zero |
| Termos raros | Peso igual | Peso elevado |
| Discriminação | Baixa | Alta |

```
# Resposta Item (6C)

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

textos = ["o sistema falhou", "o sistema foi atualizado", "falha no sistema"]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(textos)
df = pd.DataFrame(X.toarray().round(4), columns=vectorizer.get_feature_names_out(),
                  index=[f"Doc{i+1}" for i in range(len(textos))])
print("Vocabulário:", vectorizer.get_feature_names_out().tolist())
print(df.to_string())
```

## Questão 07 (1.5)

a) Explique o que é PMI (Pointwise Mutual Information) e qual é sua finalidade no processamento de linguagem natural.

b) Considerando as probabilidades abaixo, determine se “gato” e “leite” são independentes ou dependentes. Justifique sua resposta com base no conceito de independência de probabilidades:
P(gato) = 0,2
P(leite) = 0,5
P(gato, leite) = 0,1

c) Implemente em Python uma função que receba P(w, c), P(w) e P(c) e retorne o PMI correspondente.


### RESPOSTA

Item (7A)  
PMI é uma medida usada em modelos de representação vetorial baseados em coocorrência de palavras. A técnica parte de uma matriz palavra-contexto, que registra quantas vezes uma palavra aparece próxima de outra dentro de um corpus. A medida central é o PMI (Pointwise Mutual Information), que quantifica o quanto a coocorrência entre uma palavra e um contexto é maior ou menor do que seria esperado pelo acaso. Sua função é medir quão fortemente uma palavra está associada a um contexto, comparando a coocorrência observada com a coocorrência esperada ao acaso.

### RESPOSTA

Item (7B)  
Se independentes, esperaria P(gato) × P(leite) = 0,2 × 0,5 = 0,10.  
Como P(gato, leite) =0,10 é exatamente igual ao produto, as palavras são independentes (PMI = log₂(1) = 0).  
Se $P(A, B) = P(A) \times P(B)$: São independentes (não têm relação especial).  
Se $P(A, B) \neq P(A) \times P(B)$: São dependentes (existe uma associação entre elas).

```
# RESPOSTA Item (7C)
import math

def pmi(p_wc, p_w, p_c):
    if p_wc == 0 or p_w == 0 or p_c == 0:
        return float('-inf')
    return math.log2(p_wc / (p_w * p_c))

# Teste
print(pmi(0.1, 0.2, 0.5))  # Saída: 0.0 (independentes)
```

a) Explique o que representa uma matriz palavra-contexto baseada em PMI e qual seu objetivo em PLN.

b) Utilizando Python, construa uma matriz palavra-contexto com PMI para o conjunto de textos abaixo. Considere como contexto as palavras que aparecem no mesmo documento.

textos = [
    "gato bebe leite",
    "cachorro bebe água",
    "gato come peixe"
]

### RESPOSTA

8a) A matriz palavra-contexto registra, para cada par (palavra, contexto), o valor de PMI baseado na coocorrência dentro do corpus. Cada linha representa uma palavra como vetor no espaço de contextos, capturando relações semânticas latentes — palavras que aparecem em contextos semelhantes terão vetores próximos.

```
# RESPOSTA Item (8B)

# Definir um pequeno corpus para analisar coocorrência entre palavras.

corpus = [
    "gato bebe leite",
    "cachorro bebe agua",
    "gato come peixe"
]

# Visualizar o corpus.
for i, frase in enumerate(corpus, 1):
    print(f"D{i}: {frase}")

print('-'*80)


# Gerar bigramas sequenciais (janela de contexto = "1 à direita").
from nltk.util import bigrams

bigrams_list = []

for sent in corpus:
    tokens = sent.split()
    bigrams_list.extend(list(bigrams(tokens)))

bigrams_list




# Contar frequência dos bigramas.
from collections import Counter

bigram_freq = Counter(bigrams_list)
bigram_freq



# Criar matriz de coocorrência
words = sorted(set([w for bg in bigram_freq for w in bg]))

matrix = pd.DataFrame(0, index=words, columns=words)

for (w,c),freq in bigram_freq.items():
    matrix.loc[w,c] = freq

matrix


```
