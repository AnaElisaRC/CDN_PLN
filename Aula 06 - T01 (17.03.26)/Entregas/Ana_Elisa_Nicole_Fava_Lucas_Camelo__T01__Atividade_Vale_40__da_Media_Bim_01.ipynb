## Questão 01 (1.0)
a) Explique o que é uma representação esparsa em PLN e por que a codificação one-hot é considerada esparsa.

b) Na célula abaixo, implemente em Python a codificação one-hot para a lista ["casa", "carro", "filme", "bom", "ruim"] utilizando OneHotEncoder do sklearn e exiba a matriz resultante.

### RESPOSTA

Item (1A)

Trata-se de uma técnica de representação vetorial em que a grande maioria dos valores é zero. Isso ocorre porque, em um corpus com um vocabulário muito grande, cada documento utiliza apenas uma pequena parte dessas palavras. A codificação One-Hot é esparsa porque representa cada palavra como um vetor binário no qual apenas uma posição recebe o valor 1 (identificando a palavra) e todas as outras posições do vocabulário são preenchidas com 0. Como o vetor tem o tamanho de todo o vocabulário, ele acaba sendo formado quase totalmente por zeros.


```
# RESPOSTA Item (1B)
# Seu código abaixo:

from sklearn.preprocessing import OneHotEncoder

# Lista de palavras no formato 2D exigido pelo scikit-learn
palavras = [["casa"], ["carro"], ["filme"], ["bom"], ["ruim"]]

# Inicializando o encoder para retornar uma matriz densa
encoder = OneHotEncoder(sparse_output=False)

# Ajustando o vocabulário e transformando os dados na matriz one-hot
matriz_one_hot = encoder.fit_transform(palavras)

print("Vocabulário identificado e ordenado pelo modelo:", encoder.categories_[0])
print("\nMatriz Resultante:")
print(matriz_one_hot)
```

## Questão 02 (1.0)

a) Explique o que é stemming e qual é seu objetivo no processamento de linguagem natural.

b) Utilizando o NLTK e o RSLPStemmer, aplique stemming ao texto abaixo:
"Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

### RESPOSTA

Item (2A)
Stemming é uma tecnica de pré processamento na qual realiza uma redução de palavras ao seu radical aproximado, com o objetivo de reduzir as variações de palavras para melhorar a eficiência dos mecanismos de pesquisa e dos modelos de análise de texto.

```
# RESPOSTA Item (2B)
# Seu código abaixo:

import nltk
from nltk.stem import RSLPStemmer

# 1. Baixando os pacotes necessários (agora incluindo o punkt_tab)
nltk.download('punkt')
nltk.download('punkt_tab') # <-- Esta é a linha que resolve o erro!
nltk.download('rslp')

# 2. Texto original fornecido na questão
texto = "Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

# 3. Inicializando o stemmer em português
stemmer = RSLPStemmer()

# 4. Separando a frase em tokens (palavras)
tokens = nltk.word_tokenize(texto, language='portuguese')

# 5. Aplicando o stemming em cada palavra da lista
texto_stemmed = [stemmer.stem(palavra) for palavra in tokens]

# Exibindo os resultados
print("Texto original:", texto)
print("Resultado do stemming:", texto_stemmed)
```

## Questão 03 (1.0)

a) Explique o que são stopwords em PLN e qual é o objetivo de removê-las durante o processamento de texto.

b) Utilizando o NLTK, realize a remoção de stopwords do texto abaixo e exiba os tokens restantes:
"Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."

### RESPOSTA

Item (3A)

Stopwords são aquelas palavras que aparecem com muita frequência nos textos, mas que trazem pouca ou quase nenhuma informação útil para a análise, como "de", "a", "o" e "que".
O principal objetivo de remover essas palavras é padronizar e simplificar o texto, uma técnica que chamamos de normalização. Ao fazer isso, eu reduzo a complexidade dos dados (menos variações e menos palavras diferentes para analisar), o que facilita muito o processamento computacional e diminui o espaço que essas informações ocupam no modelo. Além disso, eu posso adaptar a minha lista de stopwords dependendo do assunto e do objetivo da minha tarefa

```
# RESPOSTA Item (3B)
# Seu código abaixo:
import nltk
from nltk.corpus import stopwords

# 1. Baixando os pacotes necessários para tokenização e stopwords
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# 2. Texto fornecido na questão
texto = "Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."

# 3. Separando a frase em tokens (palavras)
tokens = nltk.word_tokenize(texto, language='portuguese')

# 4. Carregando a lista padrão de stopwords em português do NLTK
stop_words = set(stopwords.words('portuguese'))

# 5. Filtrando a lista: mantém a palavra apenas se ela NÃO estiver nas stopwords.
# Usamos .lower() para garantir que palavras com letra maiúscula (como "Os") também sejam identificadas e removidas.
tokens_filtrados = [palavra for palavra in tokens if palavra.lower() not in stop_words]

# Exibindo os resultados
print("Tokens originais:", tokens)
print("\nTokens após remoção das stopwords:", tokens_filtrados)
```

### RESPOSTA

Item (5A)

Bag of Words é uma técnica de vetorização de palavras, criando um vocabulário com as palavras conhecidas e contando a frequência delas em cada frase.
Apesar de ser uma técnica simples, o BoW desconsidera o contexto e a semântica das frases, pois ignora a ordem em que as palavras aparecem.

```
# RESPOSTA Item (5B)
# Seu código abaixo:

from sklearn.feature_extraction.text import CountVectorizer

# Os textos da sua questão
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

# Inicializando o modelo BoW
vetorizador = CountVectorizer()

# Treinando o modelo e transformando os textos em uma matriz
matriz_bow = vetorizador.fit_transform(textos)

# Pegando o vocabulário criado
vocabulario = vetorizador.vocabulary_

# LINHA QUE FALTAVA: Ordenando o vocabulário pelos valores numéricos (índices)
vocabulario_ordenado = dict(sorted(vocabulario.items(), key=lambda item: item[1]))

print("Vocabulário na ordem das colunas:")
print(vocabulario_ordenado)

print("\nMatriz Resultante:")
print(matriz_bow.toarray())
```

## Questão 07 (1.5)

a) Explique o que é PMI (Pointwise Mutual Information) e qual é sua finalidade no processamento de linguagem natural.

b) Considerando as probabilidades abaixo, determine se “gato” e “leite” são independentes ou dependentes. Justifique sua resposta com base no conceito de independência de probabilidades:
P(gato) = 0,2
P(leite) = 0,5
P(gato, leite) = 0,1

c) Implemente em Python uma função que receba P(w, c), P(w) e P(c) e retorne o PMI correspondente.


### RESPOSTA

Item (7A)

PMI significa Pointwise Mutual Information (Informação Mútua Pontual). É uma medida estatística usada para descobrir a força da associação entre duas palavras. Basicamente, o PMI responde à pergunta: "Essas duas palavras aparecem juntas com mais frequência do que se fosse por puro acaso?"A fórmula matemática do PMI compara a probabilidade das duas palavras aparecerem juntas com a probabilidade de cada uma aparecer de forma independente:$$PMI(x, y) = \log_2 \left( \frac{P(x, y)}{P(x)P(y)} \right)$$O PMI é usado principalmente para encontrar colocações (collocations) e expressões compostas.

### RESPOSTA

Item (7B)

Aplicando o conceito de independência de probabilidade:

Probabilidade do gato: $P(\text{gato}) = 0.2
$Probabilidade do leite: $P(\text{leite}) = 0.5
$Probabilidade de ambos juntos: $P(\text{gato, leite}) = 0.1
$Fazendo a conta da independência:$$P(\text{gato}) \times P(\text{leite}) = 0.2 \times 0.5 = 0.1$$

Como o resultado da multiplicação ($0.1$) é exatamente igual ao $P(\text{gato, leite})$ dado no enunciado ($0.1$), nós concluímos que "gato" e "leite" são independentes.Isso significa que, nesse conjunto de textos imaginário, o fato de a palavra "gato" aparecer não aumenta nem diminui a chance de a palavra "leite" aparecer. Elas se cruzaram por puro acaso matemático. Se fossem dependentes, a probabilidade conjunta seria diferente de $0.1$.

```
# RESPOSTA Item (7C)
# Seu código abaixo:

import math

def calcular_pmi(p_wc, p_w, p_c):
    # Evitando erro matemáticos (divisão por zero ou log de zero)
    if p_w == 0 or p_c == 0 or p_wc == 0:
        return 0

    # Calculando a probabilidade independente
    prob_independente = p_w * p_c

    # Aplicando a fórmula do PMI
    pmi = math.log2(p_wc / prob_independente)

    return pmi

# --- Testando a função com os dados da letra B ---
p_gato = 0.2
p_leite = 0.5
p_ambos = 0.1

resultado_pmi = calcular_pmi(p_ambos, p_gato, p_leite)
print(f"O PMI calculado é: {resultado_pmi}")
```

## Questão 08 (1.0)

a) Explique o que representa uma matriz palavra-contexto baseada em PMI e qual seu objetivo em PLN.

b) Utilizando Python, construa uma matriz palavra-contexto com PMI para o conjunto de textos abaixo. Considere como contexto as palavras que aparecem no mesmo documento.

textos = [
    "gato bebe leite",
    "cachorro bebe água",
    "gato come peixe"
]

### RESPOSTA

Item (8A)
Significa trazer a probabilidade que uma palavra w tem de ter uma coocorrencia, ou, ser seguida de uma outra palavra que serve de contexto c.

```
import math
import pandas as pd
from collections import Counter
import itertools

textos = [
    "gato bebe leite",
    "cachorro bebe água",
    "gato come peixe"
]

# 1. Preparando os dados e calculando P(w) e P(c)
# Como o contexto é o mesmo documento, P(w) e P(c) são iguais
todas_palavras = []
for texto in textos:
    todas_palavras.extend(list(set(texto.split()))) # set() para contar 1 vez por doc

freq_palavras = Counter(todas_palavras)
N_docs = len(textos)

# 2. Encontrando os pares no mesmo documento e calculando P(w, c)
pares = []
for texto in textos:
    palavras = list(set(texto.split()))
    # itertools.combinations gera todos os pares possíveis dentro do documento
    for par in itertools.combinations(sorted(palavras), 2):
        pares.append(par)

freq_pares = Counter(pares)

# 3. Construindo a Matriz Final com o cálculo do PMI
vocabulario = sorted(list(set(todas_palavras)))
matriz_pmi = pd.DataFrame(index=vocabulario, columns=vocabulario).fillna(0.0)

for (w1, w2), coocorrencia in freq_pares.items():
    p_wc = coocorrencia / N_docs
    p_w1 = freq_palavras[w1] / N_docs
    p_w2 = freq_palavras[w2] / N_docs

    # Aplicando a fórmula do PMI: log2( P(w,c) / (P(w) * P(c)) )
    pmi = math.log2(p_wc / (p_w1 * p_w2))

    # Preenchendo a matriz espelhada
    matriz_pmi.at[w1, w2] = round(pmi, 3)
    matriz_pmi.at[w2, w1] = round(pmi, 3)

print("Matriz PMI:\n")
print(matriz_pmi)
```
