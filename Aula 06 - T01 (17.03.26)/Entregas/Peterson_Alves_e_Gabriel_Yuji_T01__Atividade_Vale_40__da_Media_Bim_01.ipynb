## Questão 01 (1.0)
a) Explique o que é uma representação esparsa em PLN e por que a codificação one-hot é considerada esparsa.

b) Na célula abaixo, implemente em Python a codificação one-hot para a lista ["casa", "carro", "filme", "bom", "ruim"] utilizando OneHotEncoder do sklearn e exiba a matriz resultante.

### RESPOSTA

Item (1A):

Em PLN, uma representação esparsa ocorre quando transformamos palavras ou documentos em vetores numéricos nos quais a grande maioria dos elementos é igual a zero. Isso acontece porque o vocabulário total de um corpus costuma ser vasto, mas cada unidade individual (como uma palavra ou uma frase curta) utiliza apenas uma fração mínima desse vocabulário. A codificação one-hot é considerada esparsa porque Cada palavra é representada por um vetor que possui o tamanho exato do vocabulário total. Nesse vetor, apenas uma única posição recebe o valor 1 (aquela que identifica a palavra em questão), enquanto todas as outras posições permanecem com 0.Se tivermos um vocabulário de 10.000 palavras, cada palavra será um vetor com 9.999 zeros e apenas um número 1, o que gera um uso de memória ineficiente e não captura relações semânticas entre os termos.

```
# RESPOSTA Item (1B)
# Seu código abaixo:

from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Lista de palavras para codificação
palavras = np.array(["casa", "carro", "filme", "bom", "ruim"]).reshape(-1, 1)

encoder = OneHotEncoder(sparse_output=False)

matriz_one_hot = encoder.fit_transform(palavras)

print("Categorias identificadas:", encoder.categories_[0])
print("\nMatriz One-Hot resultante:")
print(matriz_one_hot)
```

## Questão 02 (1.0)

a) Explique o que é stemming e qual é seu objetivo no processamento de linguagem natural.

b) Utilizando o NLTK e o RSLPStemmer, aplique stemming ao texto abaixo:
"Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

### RESPOSTA

Item (2A)

 Stemming é um tipo de normalização que consiste em "truncar" as palavras, ou seja, realizar um corte mecânico e heurístico para reduzir variações morfológicas de uma mesma base ao seu radical. Por exemplo, "estudante" e "estudar" derivam da mesma origem e, após a truncagem de seus sufixos e terminações, poderiam ser representadas pelo radical comum "estud". O objetivo principal desse processo é reduzir a dimensionalidade do vocabulário, tratando diferentes formas flexionadas de uma palavra como um único token para simplificar a análise.

```
# RESPOSTA Item (2B)
# Seu código abaixo:

import nltk
from nltk.stem import RSLPStemmer
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('rslp')

# Texto original
texto = "Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

# 1. Tokenização (Segmentação)
tokens = word_tokenize(texto, language='portuguese')

# 2. Inicialização do Stemmer (Normalização)
stemmer = RSLPStemmer()

# 3. Aplicação do stemming
stems = [stemmer.stem(token) for token in tokens]

# Exibição formatada
print(f"{'Palavra Original':<20} | {'Stem (Radical)':<15}")
print("-" * 40)
for original, stem in zip(tokens, stems):
    print(f"{original:<20} | {stem:<15}")
```

## Questão 03 (1.0)

a) Explique o que são stopwords em PLN e qual é o objetivo de removê-las durante o processamento de texto.

b) Utilizando o NLTK, realize a remoção de stopwords do texto abaixo e exiba os tokens restantes:
"Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."

### RESPOSTA

Item (3A)

Stopwords são palavras de altíssima frequência em um idioma (como artigos, preposições e conjunções) que, sozinhas, carregam pouco valor semântico para a maioria das tarefas de PLN. O objetivo de removê-las é reduzir o ruído estatístico e a dimensionalidade do vocabulário, permitindo que os modelos foquem em termos com maior poder informativo (palavras de conteúdo), o que otimiza o processamento e melhora a precisão das análises de dados textuais.

**Ex:** Gosto de uva e morango

**Removido stopwords:** gosto uva morango

```
# RESPOSTA Item (3B)
# Seu código abaixo:

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Downloads necessários para processamento em português
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# 1. Configuração do ambiente
texto = "Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."
stop_words_pt = set(stopwords.words('portuguese'))

# 2. Tokenização (Segmentação) [cite: 67]
tokens = word_tokenize(texto, language='portuguese')

# 3. Filtragem (Remoção de Stop-words)
tokens_filtrados = [w for w in tokens if w.lower() not in stop_words_pt]

# 4. Exibição dos resultados
print(f"Tokens Originais: {tokens}")
print(f"\nTokens Restantes: {tokens_filtrados}")
print(f"\nTotal de tokens removidos: {len(tokens) - len(tokens_filtrados)}")
```

## Questão 05 (1.0)

a) Explique o que é o modelo Bag of Words (BoW) e qual sua principal limitação em tarefas de PLN.

b) Utilizando o CountVectorizer do sklearn, aplique BoW ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

### RESPOSTA

Item (5A)

 O Bag of Words é um modelo de representação de texto que transforma documentos em vetores baseados na frequência de ocorrência de cada token (palavra) no vocabulário do corpus. Ele ignora completamente a estrutura gramatical e a ordem das palavras, tratando o documento apenas como uma "bolsa" de termos.

Principal Limitação: Como o BoW desconsidera a ordem, ele perde o contexto semântico. Além disso, ele gera matrizes de alta dimensionalidade e esparsas, e não diferencia palavras muito frequentes de palavras que carregam mais significado, o que pode dar o mesmo peso para termos irrelevantes.

```
# RESPOSTA Item (5B)
# Seu código abaixo:

from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Conjunto de textos fornecido
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

# 1. Inicialização do modelo Bag-of-Words
vectorizer = CountVectorizer()

# 2. Ajuste e transformação (criação da matriz de frequências)
matrix_bow = vectorizer.fit_transform(textos)

# 3. Extração do vocabulário ordenado
vocabulario = vectorizer.get_feature_names_out()

# Exibição formatada para facilitar a leitura
df_bow = pd.DataFrame(matrix_bow.toarray(), columns=vocabulario)

print("Vocabulário identificado:", vocabulario)
print("\nMatriz Bag-of-Words resultante:")
print(df_bow)
```

## Questão 06 (1.5)

a) Explique o que é TF-IDF e qual é sua principal ideia em relação à importância das palavras em um conjunto de documentos.


b) Compare TF-IDF com Bag of Words (BoW), destacando a principal diferença entre as duas abordagens.

c) Utilizando o TfidfVectorizer do sklearn, aplique TF-IDF ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:

textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

### RESPOSTA

Item (6A)

O TF-IDF é uma medida que combina quão frequente um termo é em um documento (TF) com o quão raro ele é no conjunto de documentos (IDF). Ele faz isso calculando a frequencia de uma palavra em um documento em específico em relação ao todo do corpus.

A ideia principal é que uma palavra é muito relevante se ela aparece muitas vezes em um documento (TF alto), mas essa relevância é "penalizada" se ela aparecer em quase todos os documentos do corpus (IDF baixo). Isso ajuda o modelo a identificar termos que são "assinaturas" únicas de cada texto, ignorando palavras genéricas que aparecem em todo lugar.

### RESPOSTA

Item (6B)

 A principal diferença entre as duas abordagens é como o valor de cada termo é calculado. O Bag of Words utiliza apenas a frequência absoluta das palavras em cada documento, tratando todos os termos como igualmente importantes. Já o TF-IDF aplica uma ponderação estatística que penaliza palavras muito frequentes em todo o corpus e valoriza termos mais específicos e raros que aparecem em poucos documentos. Dessa forma, o TF-IDF consegue capturar melhor a relevância semântica e informativa de cada palavra para um documento específico.

```
# RESPOSTA Item (6C)
# Seu código abaixo:

from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Conjunto de textos fornecido
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

# 1. Inicialização do TfidfVectorizer
vectorizer = TfidfVectorizer()

# 2. Ajuste e transformação (criação da matriz TF-IDF)
matrix_tfidf = vectorizer.fit_transform(textos)

# 3. Extração do vocabulário ordenado
vocabulario = vectorizer.get_feature_names_out()

# 4. Exibição da matriz em formato DataFrame para melhor visualização
df_tfidf = pd.DataFrame(matrix_tfidf.toarray(), columns=vocabulario)

print("Vocabulário identificado:", vocabulario)
print("\nMatriz TF-IDF resultante:")
print(df_tfidf.round(4)) # Arredondando para 4 casas decimais
```

## Questão 07 (1.5)

a) Explique o que é PMI (Pointwise Mutual Information) e qual é sua finalidade no processamento de linguagem natural.

b) Considerando as probabilidades abaixo, determine se “gato” e “leite” são independentes ou dependentes. Justifique sua resposta com base no conceito de independência de probabilidades:
P(gato) = 0,2
P(leite) = 0,5
P(gato, leite) = 0,1

c) Implemente em Python uma função que receba P(w, c), P(w) e P(c) e retorne o PMI correspondente.



### RESPOSTA

Item (7A)

O PMI é uma medida de associação estatística que quantifica a dependência entre duas palavras. Sua finalidade no PLN é identificar se um par de termos ocorre junto com uma frequência significativamente maior do que o esperado se ambos fossem independentes.

Em outras palavras, o PMI nos diz se a presença de uma palavra aumenta a probabilidade de a outra palavra estar presente no mesmo contexto. Se o valor for positivo e alto, indica uma forte relação semântica ou uma colocalização. Se for próximo de zero, as palavras ocorrem juntas apenas por mero acaso estatístico.

### RESPOSTA

Item (7B)

De acordo com o conceito de independência em probabilidade, dois eventos são considerados independentes quando a probabilidade da ocorrência de ambos juntos é igual à multiplicação de suas probabilidades individuais.

Aplicando aos dados fornecidos:

A probabilidade de "gato" é 0,2.
A probabilidade de "leite" é 0,5.
A probabilidade conjunta de "gato" e "leite" juntos é 0,1.
Ao multiplicar as probabilidades individuais (0,2 x 0,5), obtemos o resultado 0,1. Como esse valor é exatamente igual à probabilidade conjunta informada no exercício (0,1), as palavras "gato" e "leite" são classificadas como estatisticamente independentes. Isso significa que, neste conjunto de dados, o aparecimento de uma dessas palavras não tem relação direta com o aparecimento da outra; elas ocorrem juntas apenas por um mero acaso estatístico, conforme comprovado no código da questão 7C.

```
# RESPOSTA Item (7C)
# Seu código abaixo:

import numpy as np

def calcular_pmi(p_wc, p_w, p_c):
    """
    Função para calcular o PMI (Pointwise Mutual Information).
    Fórmula: PMI = log2( P(w,c) / (P(w) * P(c)) )
    """
    # Calculamos a razão entre a probabilidade conjunta e o produto das individuais
    ratio = p_wc / (p_w * p_c)

    # Calculamos o logaritmo na base 2
    pmi = np.log2(ratio)

    return pmi

# Teste com os dados da Questão 7b:
p_gato = 0.2
p_leite = 0.5
p_conjunta = 0.1

resultado = calcular_pmi(p_conjunta, p_gato, p_leite)

print(f"O PMI entre 'gato' e 'leite' é: {resultado}")
```

## Questão 08 (1.0)

a) Explique o que representa uma matriz palavra-contexto baseada em PMI e qual seu objetivo em PLN.

b) Utilizando Python, construa uma matriz palavra-contexto com PMI para o conjunto de textos abaixo. Considere como contexto as palavras que aparecem no mesmo documento.

textos = [
    "gato bebe leite",
    "cachorro bebe água",
    "gato come peixe"
]

### RESPOSTA

Item (8A)

Uma matriz palavra-contexto baseada em PMI representa as palavras como vetores numéricos, onde cada linha é uma palavra alvo e cada coluna é um contexto possível. Diferente de uma matriz de contagem bruta, os valores de cada célula são o PMI entre a palavra e o contexto.

A finalidade dessa matriz é capturar a semântica das palavras. Através do PMI, o modelo identifica relações de significado: palavras que ocorrem frequentemente nos mesmos contextos estatísticos (mesmo que nunca apareçam juntas no mesmo texto) tendem a ter vetores similares, o que nos permite medir a proximidade de significado entre termos diferentes.

```
import pandas as pd
import numpy as np
from itertools import combinations

# 1. Preparação do Corpus e Vocabulário
textos = ["gato bebe leite", "cachorro bebe água", "gato come peixe"]
docs = [doc.split() for doc in textos]
vocab = sorted(list(set(word for doc in docs for word in doc)))

# 2. Construção da Matriz de Coocorrência
cooc = pd.DataFrame(0, index=vocab, columns=vocab, dtype=float)
for doc in docs:
    # Geramos todas as combinações de pares de palavras no mesmo documento
    for w1, w2 in combinations(doc, 2):
        cooc.loc[w1, w2] += 1
        cooc.loc[w2, w1] += 1

# 3. Cálculo das Probabilidades
N = cooc.values.sum()
p_wc = cooc / N
p_w = p_wc.sum(axis=1)
p_c = p_wc.sum(axis=0)

# 4. Cálculo do PMI (Base 2 para seguir o padrão de bits do PLN)
with np.errstate(divide='ignore', invalid='ignore'):
    pmi_matrix = np.log2(p_wc / np.outer(p_w, p_c))

# Limpeza: PPMI (Positive PMI) - mantemos apenas associações positivas
pmi_matrix = pmi_matrix.replace([np.inf, -np.inf], 0).clip(lower=0).fillna(0)

print("Matriz Palavra-Contexto (Valores de PMI):")
print(pmi_matrix.round(3))
```
