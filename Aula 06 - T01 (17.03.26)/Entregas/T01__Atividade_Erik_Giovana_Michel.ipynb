## Questão 01 (1.0)
a) Explique o que é uma representação esparsa em PLN e por que a codificação one-hot é considerada esparsa.

b) Na célula abaixo, implemente em Python a codificação one-hot para a lista ["casa", "carro", "filme", "bom", "ruim"] utilizando OneHotEncoder do sklearn e exiba a matriz resultante.

### RESPOSTA

Item (1A)
Uma representação esparsa em PLN refere-se a vetores numéricos de alta dimensionalidade onde a imensa maioria dos valores é preenchida com zero
Isso acontece porque a dimensão do vetor é definida pelo tamanho do vocabulário total do corpus, mas cada documento individual utiliza apenas uma fração muito pequena desse total de palavras, deixando o resto das posições vazias (zeradas)

A codificação one-hot é o exemplo mais básico de representação esparsa porque converte cada palavra (token) em um vetor binário do exato tamanho do vocabulário, no qual apenas uma única posição recebe o valor 1 (aquela que identifica a palavra) e absolutamente todas as outras posições recebem o valor 0.

```python
# RESPOSTA Item (1B)
# Seu código abaixo:

from sklearn.preprocessing import OneHotEncoder
import numpy as np

# A entrada para o OneHotEncoder deve ser uma matriz 2D
palavras = [["casa"], ["carro"], ["filme"], ["bom"], ["ruim"]]

# Inicializa o encoder. O parâmetro sparse_output=False retorna um array NumPy normal (denso) para facilitar a visualização
encoder = OneHotEncoder(sparse_output=False)

# Ajusta e transforma os dados
matriz_onehot = encoder.fit_transform(palavras)

# Exibe os resultados
print("Categorias:", encoder.categories_)
print("Matriz One-Hot Resultante:\n", matriz_onehot)
```

## Questão 02 (1.0)

a) Explique o que é stemming e qual é seu objetivo no processamento de linguagem natural.

b) Utilizando o NLTK e o RSLPStemmer, aplique stemming ao texto abaixo:
"Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

### RESPOSTA

Item (2A)
O stemming é basicamente uma forma de simplificar palavras. Ele pega uma palavra e corta ela até chegar no “pedaço principal”, que é o radical.

A ideia aqui não é ser perfeito na gramática nem no sentido da palavra, é mais um corte mesmo, meio direto. Ele vai tirando final, começo, essas partes extras, usando algumas regras prontas (tipo aqueles algoritmos já definidos pra isso).

Por exemplo: “correndo”, “correu” e “corrida” acabam virando tudo “corr”.

```python
# RESPOSTA Item (2B)
# Seu código abaixo:
import nltk
from nltk.stem import RSLPStemmer
from nltk.tokenize import word_tokenize


nltk.download('rslp')
nltk.download('punkt')
nltk.download('punkt_tab')

texto = "Os pesquisadores analisavam relatórios e comparavam resultados financeiros."

stemmer = RSLPStemmer()

# Tokenização corrigida
tokens = word_tokenize(texto, language='portuguese')

resultado = [stemmer.stem(palavra) for palavra in tokens]

print("Original:", tokens)
print("Com Stemming:", " ".join(resultado))
```

## Questão 03 (1.0)

"Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."

### RESPOSTA

Item (3A)
Stopwords são palavras muito comuns no idioma, como artigos, preposições e conjunções (por exemplo: “o”, “a”, “de”, “e”), que geralmente não trazem muito significado relevante sozinhas em um texto no contexto de PLN.

A ideia de removê-las é deixar o texto mais “limpo”, tirando essas palavras que não agregam tanto, para reduzir o volume de dados e facilitar o processamento. Com isso, ficam em destaque as palavras mais importantes, o que ajuda a melhorar a eficiência e o desempenho em tarefas como classificação de textos, busca de informações e análise de sentimentos.



```python
# RESPOSTA Item (3B)
# Seu código abaixo:
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# baixar recursos (caso ainda não tenha)
nltk.download('punkt')
nltk.download('stopwords')

texto = "Os pesquisadores analisavam os dados e comparavam os resultados com os modelos anteriores."

# tokenização
tokens = word_tokenize(texto, language='portuguese')

# stopwords em português
stop_words = set(stopwords.words('portuguese'))

# remoção das stopwords
tokens_filtrados = [palavra for palavra in tokens if palavra.lower() not in stop_words]

print(tokens_filtrados)
```

## Questão 05 (1.0)

a) Explique o que é o modelo Bag of Words (BoW) e qual sua principal limitação em tarefas de PLN.

b) Utilizando o CountVectorizer do sklearn, aplique BoW ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

### RESPOSTA

Item (5A)
O Bag‑of‑Words é uma forma de representar textos transformando cada documento em um vetor baseado apenas na frequência das palavras. Cada posição do vetor corresponde a um termo do vocabulário, e o valor indica quantas vezes ele aparece no documento.
limitações importantes:
- Ignora ordem das palavras, sintaxe e estrutura gramatical.
- Não considera contexto, o que dificulta lidar com ambiguidades.
- Palavras com múltiplos sentidos recebem a mesma representação, como “banco”, "manga" (instituição financeira ou assento, camisa ou fruta), prejudicando a compreensão semântica.


```python
# RESPOSTA Item (5B)
# Seu código abaixo:
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Conjunto de textos
textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

# Inicializa o CountVectorizer (por padrão, ele converte para minúsculas e remove pontuação)
vectorizer = CountVectorizer()

# Ajusta o modelo aos textos e transforma em uma matriz BoW
matriz_bow = vectorizer.fit_transform(textos)

# Obtém o vocabulário gerado (as colunas da matriz)
vocabulario = vectorizer.get_feature_names_out()

# Cria um DataFrame para exibir a matriz de forma legível
df_bow = pd.DataFrame(matriz_bow.toarray(), columns=vocabulario, index=["Texto 1", "Texto 2", "Texto 3"])

print("Vocabulário:", vocabulario)
print("\nMatriz Resultante Bag-of-Words:")
print(df_bow)
```

## Questão 06 (1.5)

a) Explique o que é TF-IDF e qual é sua principal ideia em relação à importância das palavras em um conjunto de documentos.

b) Compare TF-IDF com Bag of Words (BoW), destacando a principal diferença entre as duas abordagens.

c) Utilizando o TfidfVectorizer do sklearn, aplique TF-IDF ao conjunto de textos abaixo e exiba a matriz resultante e o vocabulário:

textos = [
    "o sistema falhou",
    "o sistema foi atualizado",
    "falha no sistema"
]

### RESPOSTA

Item (6A)
O TF-IDF é uma forma de transformar texto em números, mas de um jeito mais inteligente. Ele basicamente mede o quanto uma palavra é importante dentro de um texto, comparando com vários outros textos.

A lógica é simples:
se uma palavra aparece muito em um documento, provavelmente ela é importante ali. Só que tem um detalhe — se essa mesma palavra aparece em praticamente todos os textos (tipo “o”, “e”, “que”), então ela não ajuda muito a diferenciar nada.

Então o TF-IDF faz esse equilíbrio: ele valoriza palavras que aparecem bastante em um texto, mas que não são tão comuns nos outros.

No fim, isso ajuda o modelo a entender melhor sobre o que aquele texto realmente fala.

### RESPOSTA

Item (6B)
O BoW e o TF-IDF fazem basicamente a mesma coisa: transformar texto em número pra máquina conseguir entender. A diferença tá em como cada um enxerga o que é importante numa palavra.

O Bag of Words (BoW) é o mais simples de todos. Ele funciona tipo um contador mesmo.
Ele pega todas as palavras que existem nos textos e, pra cada documento, só conta quantas vezes cada palavra aparece.

O problema é que ele não filtra nada. Então, se uma palavra tipo “o” ou “e” aparece 50 vezes, ele vai achar que aquilo é super importante — sendo que não diz nada sobre o conteúdo.

Já o TF-IDF é uma versão mais inteligente disso.
Ele também conta as palavras, mas faz um ajuste: ele diminui o peso das palavras que aparecem em todo lugar e aumenta o peso das que são mais específicas daquele texto.

Ou seja, ele não olha só frequência — ele olha o quanto aquela palavra realmente ajuda a diferenciar um texto dos outros.

No fim, dá pra resumir assim:
o BoW só conta… o TF-IDF tenta entender o que realmente importa.

```python
# RESPOSTA Item (6C)
# Seu código abaixo:
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1. Definir o conjunto de textos
textos = [ "o sistema falhou", "o sistema foi atualizado", "falha no sistema" ]

# 2. Instanciar o Vetorizador
# Nota: Por padrão, o TfidfVectorizer remove pontuação e coloca tudo em minúsculo
vectorizer = TfidfVectorizer()

# 3. Ajustar e transformar os textos em uma matriz TF-IDF
tfidf_matrix = vectorizer.fit_transform(textos)

# 4. Exibir o Vocabulário (Mapeamento de cada palavra a um índice na matriz)
print("Vocabulário:", vectorizer.vocabulary_)

# 5. Exibir a Matriz Resultante (Convertida para DataFrame para facilitar a leitura)
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=["Doc 1", "Doc 2", "Doc 3"]
)

print("\nMatriz TF-IDF:")
print(df_tfidf)
```

## Questão 07 (1.5)

a) Explique o que é PMI (Pointwise Mutual Information) e qual é sua finalidade no processamento de linguagem natural.

b) Considerando as probabilidades abaixo, determine se “gato” e “leite” são independentes ou dependentes. Justifique sua resposta com base no conceito de independência de probabilidades:
P(gato) = 0,2
P(leite) = 0,5
P(gato, leite) = 0,1

c) Implemente em Python uma função que receba P(w, c), P(w) e P(c) e retorne o PMI correspondente.


### RESPOSTA

Item (7A)
PMI (Pointwise Mutual Information) é uma medida estatística que indica o quanto duas palavras estão associadas, ou seja, se elas aparecem juntas com mais frequência do que seria esperado ao acaso.

No contexto de PLN, a finalidade do PMI é identificar relações importantes entre palavras em um texto ou conjunto de textos. Ele é usado para analisar coocorrência, ajudando a descobrir palavras que têm uma forte ligação entre si, como expressões ou termos que costumam aparecer juntos.

Assim, o PMI é útil em tarefas como extração de termos relevantes e identificação de associações entre palavras. Em resumo, ele serve para medir a força da relação entre palavras e destacar conexões importantes no texto.

### RESPOSTA

Item (7B)
Para verificar se “gato” e “leite” são independentes teríanos que usar a definição de independência de probabilidades:

Dois eventos são independentes quando: ( P(gato, leite) = P(gato) \times P(leite) )

Substituindo os valores: ( P(gato) \times P(leite) = 0,2 \times 0,5 = 0,1 )

Como o valor encontrado (0,1) é igual a ( P(gato, leite) = 0,1 ), podemos concluir que os eventos são independentes. Ou seja, a ocorrência da palavra “gato” não influencia a ocorrência da palavra “leite”, pois a probabilidade conjunta é exatamente igual ao produto das probabilidades individuais.

os eventos são perfeitamente independentes

```python
# RESPOSTA Item (7C)
# Seu código abaixo:

#Para calcular o PMI em Python, podemos criar uma função que recebe as probabilidades

import math

def calcular_pmi(p_wc, p_w, p_c):
    return math.log2(p_wc / (p_w * p_c))


pmi = calcular_pmi(0.1, 0.2, 0.5)
print(pmi)
```

## Questão 08 (1.0)

a) Explique o que representa uma matriz palavra-contexto baseada em PMI e qual seu objetivo em PLN.

b) Utilizando Python, construa uma matriz palavra-contexto com PMI para o conjunto de textos abaixo. Considere como contexto as palavras que aparecem no mesmo documento.

textos = [
    "gato bebe leite",
    "cachorro bebe água",
    "gato come peixe"
]

### RESPOSTA

Item (8A)
Uma matriz palavra‑contexto representa cada palavra por meio das coocorrências com outras palavras que aparecem ao seu redor no corpus. As linhas correspondem às palavras e as colunas aos contextos, registrando quantas vezes cada combinação ocorre.
Quando essa matriz usa PMI (Pointwise Mutual Information), ela deixa de trabalhar com contagens brutas e passa a medir o quanto a associação entre palavra e contexto é maior (ou menor) do que o esperado ao acaso.

O objetivo dessa abordagem é capturar relações semânticas latentes: palavras que compartilham contextos semelhantes tendem a ter significados próximos. Assim, a similaridade deixa de depender da repetição literal das mesmas palavras (como em BoW ou TF‑IDF) e passa a refletir afinidades estatísticas que revelam proximidade semântica.


```python
# RESPOSTA Item (8B)
# Seu código abaixo:
import numpy as np
import pandas as pd
import math
from collections import Counter

textos = [
    "gato bebe leite",
    "cachorro bebe água",
    "gato come peixe"
]

# 1. Tokenização simples (separação por espaços)
docs = [texto.split() for texto in textos]

# 2. Obter o vocabulário único
vocabulario = list(set([palavra for doc in docs for palavra in doc]))

# 3. Contagem de coocorrências e frequências individuais
cooc_counts = Counter()
freq_w = Counter()
freq_c = Counter()

# Como o contexto é o "mesmo documento", cada palavra pareia com todas as outras do documento
for doc in docs:
    for i, w in enumerate(doc):
        for j, c in enumerate(doc):
            if i != j: # A palavra não coocorre com ela mesma na mesma posição
                cooc_counts[(w, c)] += 1
                freq_w[w] += 1
                freq_c[c] += 1

# N = número total de coocorrências consideradas no corpus
N = sum(cooc_counts.values())

# 4. Construção da Matriz PMI
pmi_matrix = pd.DataFrame(index=vocabulario, columns=vocabulario, dtype=float)

for w in vocabulario:
    for c in vocabulario:
        cooc = cooc_counts[(w, c)]

        # Apenas calcula o PMI se as palavras coocorrem ao menos uma vez para evitar log(0)
        if cooc > 0:
            # P(w,c) = probabilidade conjunta
            p_wc = cooc / N
            # P(w) e P(c) = probabilidades marginais
            p_w = freq_w[w] / N
            p_c = freq_c[c] / N

            # Fórmula do PMI: log( P(w,c) / (P(w) * P(c)) )
            pmi = math.log(p_wc / (p_w * p_c))
            pmi_matrix.at[w, c] = round(pmi, 4)
        else:
            pmi_matrix.at[w, c] = np.nan # Valores nulos onde não há coocorrência

print("Matriz Palavra-Contexto baseada em PMI:")
print(pmi_matrix)
```
