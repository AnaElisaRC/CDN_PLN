# Prática — Módulo 4 – Representação Vetorial Baseada em Coocorrência

## Prática
Nesta prática o aluno constrói passo a passo diferentes representações vetoriais a partir de um pequeno corpus textual. Inicialmente são exploradas estatísticas básicas do corpus, seguida da geração de matrizes BoW e TF-IDF. Em seguida são extraídos bigramas sequenciais para construir uma matriz de coocorrência palavra-contexto. A partir dessa matriz são calculadas probabilidades conjuntas, probabilidades marginais e, por fim, os valores de PMI e PPMI, permitindo observar como relações estatísticas entre palavras emergem diretamente dos dados.

## Dataset
O corpus utilizado consiste em um conjunto pequeno de frases sintéticas envolvendo agentes (como gato, cachorro e homem), verbos de ação (bebe, come) e alimentos ou líquidos (leite, água, carne, peixe). Embora simples, esse corpus é suficiente para ilustrar como padrões de coocorrência podem revelar associações entre palavras, como verbos e objetos, permitindo visualizar o funcionamento das matrizes de coocorrência e das medidas de associação utilizadas em PLN.




```python
# Importar bibliotecas
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
# Definir um pequeno corpus para analisar coocorrência entre palavras.

# Corpus minúsculo.
corpus = [
    "gato bebe agua",
    "cachorro bebe agua",
    "gato bebe leite",
    "cachorro bebe leite",
    "gato come peixe",
    "cachorro come peixe",
    "gato come carne",
    "cachorro come carne",
    "homem bebe agua",
    "homem come carne",
]

corpus

# Visualizar o corpus.
for i, frase in enumerate(corpus, 1):
    print(f"D{i}: {frase}")
```

## PARTE 3 — Coocorrência (n-gramas)

```python
# Gerar bigramas sequenciais (janela de contexto = "1 à direita").
from nltk.util import bigrams

bigrams_list = []

for sent in corpus:
    tokens = sent.split()
    bigrams_list.extend(list(bigrams(tokens)))

bigrams_list
```

```python
# Contar frequência dos bigramas.
from collections import Counter

bigram_freq = Counter(bigrams_list)
bigram_freq
```

```python
# Criar matriz de coocorrência a partir dos bigramas.
words = sorted(set([w for bg in bigram_freq for w in bg]))

matrix = pd.DataFrame(0, index=words, columns=words)

for (w,c),freq in bigram_freq.items():
    matrix.loc[w,c] = freq

matrix
```

## PARTE 4 — PPMI

### Pergunta
15. O que significa a probabilidade conjunta P(w,c) para um par de palavras?
    R. A probabilidade conjunta P(w,c) indica a probabilidade de dois eventos ocorrerem ao mesmo tempo, ou seja, de um par específico de palavras (uma palavra e o seu contexto) ocorrer em relação ao total de todos os pares observados no corpus
    Matematicamente, ela é calculada dividindo o número de vezes que essas duas palavras aparecem juntas (cooc(w,c)) pelo número total de ocorrências consideradas (N) no corpus



```python
# Calcular total de coocorrências.
N = matrix.values.sum()
N
```

```python
# Calcular probabilidade conjunta dos bigramas.
N = sum(bigram_freq.values())

p_wc = {bg: freq / N for bg, freq in bigram_freq.items()}
p_wc
```

### Pergunta
16. Qual é o valor da soma das probabilidades conjuntas? O resultado seria o mesmo independentemente da quantidade de palavras?
    R. Sim, o resultado seria exatamente o mesmo independentemente da quantidade de palavras no vocabulário ou do tamanho do corpus. Isso ocorre porque a probabilidade conjunta P(w,c) é calculada dividindo o número de vezes que as palavras aparecem juntas (cooc(w,c)) pelo número total de ocorrências consideradas (N)
    Como a soma de todas as coocorrências possíveis na matriz formará exatamente o valor total de ocorrências (N), a razão estatística final será sempre N/N=1, independentemente de quantas palavras o texto possua


```python
# Gerar matrix de probabilidade conjunta.
N = matrix.values.sum()

p_wc = matrix / N
p_wc
```

### Pergunta
17. Se P(w,c) é alto, por si só, isso implica forte associação entre w e c?

    R. Não. Um valor alto de probabilidade conjunta P(w,c) (frequência com que aparecem juntas) não garante, por si só, que exista uma forte afinidade semântica entre as palavras
    Isso ocorre porque palavras muito frequentes de forma geral no corpus (como verbos muito comuns ou conectivos) tendem a registrar altas contagens de coocorrência com quase qualquer outra palavra do texto
    Consequentemente, essa alta ocorrência conjunta pode ser apenas um reflexo da frequência isolada de cada palavra, e utilizar apenas a contagem bruta pode acabar mascarando as associações semânticas que são realmente específicas e interessantes
18. Como saber se duas palavras aparecem juntas mais do que seria esperado ao acaso?

    R. Para descobrir isso, utiliza-se a medida estatística PMI (Pointwise Mutual Information)
    O PMI não avalia apenas a contagem de vezes que as palavras aparecem juntas; ele compara a coocorrência observada (probabilidade conjunta) com a coocorrência que seria esperada caso as palavras fossem eventos totalmente independentes (o acaso)


```python
# Calcular frequência marginal das palavras.
p_w = matrix.sum(axis=1) / N
p_w
```

```python
# Calcular frequência marginal dos contextos.
p_c = matrix.sum(axis=0) / N
p_c
```

```python
# Calcular PMI (Pointwise Mutual Information).
import numpy as np

pmi = np.log(p_wc / (p_w.values[:, None] * p_c.values[None, :]))
pmi = pmi.replace([np.inf, -np.inf], 0)
pmi
```

### Pergunta
19. Porque apareceu NaN?

    R. por que o resultado deu erro matematico, nao sendo possivel armazenar como numero, exemplo divisão por zero
20. O que acontece no passo chamado Positive PMI?

    R. O que acontece no passo chamado Positive PMI? No passo do PPMI (Positive Pointwise Mutual Information), o algoritmo resolve o problema do PMI substituindo todos os valores negativos e matematicamente indefinidos por zero

```python
# Converter para PPMI.
ppmi = pmi.clip(lower=0).fillna(0)
ppmi
```
