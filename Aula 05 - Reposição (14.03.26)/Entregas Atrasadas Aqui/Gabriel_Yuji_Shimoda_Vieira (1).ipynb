# Prática — Módulo 4 – Representação Vetorial Baseada em Coocorrência

## Prática
Nesta prática o aluno constrói passo a passo diferentes representações vetoriais a partir de um pequeno corpus textual. Inicialmente são exploradas estatísticas básicas do corpus, seguida da geração de matrizes BoW e TF-IDF. Em seguida são extraídos bigramas sequenciais para construir uma matriz de coocorrência palavra-contexto. A partir dessa matriz são calculadas probabilidades conjuntas, probabilidades marginais e, por fim, os valores de PMI e PPMI, permitindo observar como relações estatísticas entre palavras emergem diretamente dos dados.

## Dataset
O corpus utilizado consiste em um conjunto pequeno de frases sintéticas envolvendo agentes (como gato, cachorro e homem), verbos de ação (bebe, come) e alimentos ou líquidos (leite, água, carne, peixe). Embora simples, esse corpus é suficiente para ilustrar como padrões de coocorrência podem revelar associações entre palavras, como verbos e objetos, permitindo visualizar o funcionamento das matrizes de coocorrência e das medidas de associação utilizadas em PLN.




```
# Importar bibliotecas
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```
# Definir um pequeno corpus para analisar coocorrência entre palavras.

# Corpus minúsculo.
corpus = [
    "gato bebe agua",
    "cachorro bebe agua",
    "gato bebe leite",
    "cachorro bebe leite",
    "gato come peixe",
    "cachorro come peixe",
    "gato come carne",
    "cachorro come carne",
    "homem bebe agua",
    "homem come carne",
]

corpus

# Visualizar o corpus.
for i, frase in enumerate(corpus, 1):
    print(f"D{i}: {frase}")
```

## PARTE 1 — BoW

### Pergunta
1. Quantas palavras diferentes existem no corpus?

R:Existem 9 palavras diferentes no total (água, bebe, cachorro, carne, come, gato, homem, leite, peixe). Esse é o tamanho do vocabulário.

2. Quais palavras aparecem em mais de um documento?

R:Todas as 9 palavras aparecem em pelo menos dois documentos (frases) diferentes.

```
# Contar palavras no corpus.
from collections import Counter

tokens = " ".join(corpus).split()

freq = Counter(tokens)
freq.most_common()
```

```
# Criar matriz Bag of Words.
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(corpus)

bow = pd.DataFrame(
    X_bow.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"D{i+1}" for i in range(len(corpus))]
)

bow
```

### Pergunta
3. Quantos vetores existem na matriz BoW?

R:Existem 10 vetores, pois o corpus possui 10 documentos (frases).

4. Cada vetor representa o quê?

R:Cada vetor representa um documento inteiro (uma frase) do seu corpus, traduzido para números com base na contagem de palavras.

5. Qual é a dimensão de cada vetor?

R:A dimensão é 9, correspondente ao número total de palavras únicas no vocabulário.

6. A dimensão do vetor depende de quê?

R:Depende exclusivamente do tamanho do vocabulário, ou seja, da quantidade de palavras únicas presentes em todo o corpus analisado.

7. Documentos que compartilham palavras terão vetores parecidos?

R:Sim. Se duas frases têm palavras em comum, seus vetores terão valores (1s) nas mesmas colunas (posições), tornando-os matematicamente mais próximos.

8. Pela matriz BoW, é possível perceber que “bebe” aparece tanto com “gato” quanto com “cachorro”? Ou essa relação não fica explícita na matriz?

R:Não fica explícito como uma relação de pares. O BoW apenas diz que a palavra "bebe" está no Documento 1 e a palavra "gato" também está no Documento 1. Ele modela a relação Documento $\rightarrow$ Palavra, e não a relação direta de coocorrência Palavra $\rightarrow$ Palavra.

PARTE 2 — TF-IDF

### Pergunta
9. A matriz TF-IDF possui o mesmo número de vetores do BoW?

R:Sim, continuam sendo 10 vetores, um para cada frase.

10. A dimensão dos vetores mudou em relação ao BoW?

R:Não, a dimensão continua sendo 9 (o tamanho do vocabulário).

11. Por que agora aparecem valores decimais em vez de contagens inteiras?

R:Porque o TF-IDF não apenas conta as palavras, ele calcula um peso. Ele pega a frequência da palavra na frase e penaliza (diminui o peso) se essa palavra for muito comum em todo o corpus, além de aplicar uma normalização matemática para equilibrar o tamanho dos vetores.

12. Todas as palavras possuem o mesmo peso dentro de um documento?

R:Não. Palavras mais raras no corpus geral (como "leite" ou "homem") recebem pesos maiores do que palavras muito comuns (como "bebe" ou "come").

13. Qual é a palavra mais informativa (mais discriminativa) de cada documento?

R:É a palavra que possui o maior valor decimal no vetor daquele documento. Por exemplo, na frase "homem bebe água", "homem" e "água" terão pesos maiores que "bebe", pois ajudam mais a distinguir essa frase das demais.

```
# gerar matriz TF-IDF.
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(corpus)

tfidf = pd.DataFrame(
    X_tfidf.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"D{i+1}" for i in range(len(corpus))]
)

tfidf
```

### Pergunta
14. O TF-IDF consegue capturar relação entre palavras como bebe → leite?

R:Não. Assim como o BoW, o TF-IDF descarta a ordem das palavras e a sintaxe da frase. Ele avalia apenas a importância da palavra isolada para o documento.

## PARTE 3 — Coocorrência (n-gramas)

```
# Gerar bigramas sequenciais (janela de contexto = "1 à direita").
from nltk.util import bigrams

bigrams_list = []

for sent in corpus:
    tokens = sent.split()
    bigrams_list.extend(list(bigrams(tokens)))

bigrams_list
```

```
# Contar frequência dos bigramas.
from collections import Counter

bigram_freq = Counter(bigrams_list)
bigram_freq
```

```
# Criar matriz de coocorrência a partir dos bigramas.
words = sorted(set([w for bg in bigram_freq for w in bg]))

matrix = pd.DataFrame(0, index=words, columns=words)

for (w,c),freq in bigram_freq.items():
    matrix.loc[w,c] = freq

matrix
```

## PARTE 4 — PPMI

### Pergunta
15. O que significa a probabilidade conjunta P(w,c) para um par de palavras?

R:Significa a chance de essas duas palavras aparecerem coladas (uma sendo contexto da outra) se você sortear um bigrama aleatório de todo o corpus.


```
# Calcular total de coocorrências.
N = matrix.values.sum()
N
```

```
# Calcular probabilidade conjunta dos bigramas.
N = sum(bigram_freq.values())

p_wc = {bg: freq / N for bg, freq in bigram_freq.items()}
p_wc
```

### Pergunta
16. Qual é o valor da soma das probabilidades conjuntas? O resultado seria o mesmo independentemente da quantidade de palavras?

R:A soma é 1 (ou 100%). Sim, o resultado seria o mesmo independentemente do tamanho do corpus, pois estamos lidando com uma distribuição de probabilidade fechada.

```
# Gerar matrix de probabilidade conjunta.
N = matrix.values.sum()

p_wc = matrix / N
p_wc
```

### Pergunta
17. Se P(w,c) é alto, por si só, isso implica forte associação entre w e c?

R:Não necessariamente. Se "bebe" e "água" aparecem muitas vezes no texto de forma geral, é natural que a coocorrência delas também seja alta apenas pelo acaso, e não obrigatoriamente por uma ligação semântica super forte.

18. Como saber se duas palavras aparecem juntas mais do que seria esperado ao acaso?

R:Você compara a chance de elas aparecerem juntas (P(w,c)) com a chance de elas aparecerem de forma independente multiplicadas (P(w) x P(c)). É exatamente esse cálculo que o PMI faz.


```
# Calcular frequência marginal das palavras.
p_w = matrix.sum(axis=1) / N
p_w
```

```
# Calcular frequência marginal dos contextos.
p_c = matrix.sum(axis=0) / N
p_c
```

```
# Calcular PMI (Pointwise Mutual Information).
import numpy as np

pmi = np.log(p_wc / (p_w.values[:, None] * p_c.values[None, :]))
pmi = pmi.replace([np.inf, -np.inf], 0)
pmi
```

### Pergunta
19. Porque apareceu NaN?

R:"NaN" (Not a Number) aparece porque a fórmula do PMI utiliza um logaritmo: $\log(P(w,c) \dots)$. Quando duas palavras nunca aparecem juntas, o valor de coocorrência é zero. Como o logaritmo de zero não existe (tende a menos infinito), a biblioteca do Python retorna um erro matemático ou NaN.

20. O que acontece no passo chamado Positive PMI?

R:O PPMI "limpa" a matriz do PMI. Ele transforma todos os valores negativos e os "NaN" em 0. Assim, a matriz passa a destacar apenas as palavras que têm associações positivas (que aparecem juntas mais do que o acaso).

```
# Converter para PPMI.
ppmi = pmi.clip(lower=0).fillna(0)
ppmi
```

### Pergunta
21. A matriz PPMI indica claramente quais alimentos estão associados a bebe e quais estão associados a come. No entanto, é possível determinar a partir dela qual agente (gato, cachorro, homem) bebe ou come cada alimento específico?

R:Não é possível. Como usamos uma "janela" de apenas 1 palavra à direita (bigramas), a matriz ligou o Sujeito ao Verbo ("gato" $\rightarrow$ "bebe") e o Verbo ao Objeto ("bebe" $\rightarrow$ "água"). A conexão direta "gato" $\rightarrow$ "água" se perdeu.

22. Esse tipo de identificação de associação entre verbos e objetos (por exemplo, bebe–água ou come–peixe) era possível utilizando apenas BoW ou TF-IDF? Explique por quê.

R:Não. O BoW e o TF-IDF tratam as frases como um "saco de palavras" desordenado. Eles apenas sabem quais palavras estão presentes na frase, mas destroem completamente a ordem e a estrutura de vizinhança. Para associar verbos a objetos estatisticamente, é necessário olhar para a ordem, que é o que os bigramas e matrizes de coocorrência fazem.
